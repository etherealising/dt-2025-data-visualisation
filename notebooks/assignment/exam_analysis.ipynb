```python
import pandas as pd

df=pd.read_csv("student_exam_scores.csv")
df.head()
```

Does studying affect students' sleep?

Studying does not necessarily affect students' sleep, as some students study for many hours and sleep a while, while some sleep less.

```python
import matplotlib.pyplot as plt

avg_hours=df.groupby("hours_studied")["sleep_hours"].mean()
avg_hours.plot(kind="bar")
plt.title("Hours slept vs. Hours studied")
plt.ylabel("Hours slept")
plt.xlabel("Hours studied")
plt.show()
```

Do hours studied affect exam scores?

The more a student studies, the more likely they are to score higher.

```python
avg_score=df.groupby("hours_studied")["exam_score"].mean().sort_values(ascending=False)
avg_score.plot(kind="bar")
plt.title("Average score per hours studied")
plt.ylabel("Average score")
plt.xlabel("Hours studied")
plt.show()
```

```python
avg_grades=df.groupby("hours_studied")["exam_score"].mean().sort_values(ascending=False)
plt.figure(figsize=(10, 10))
avg_grades.plot(kind="pie", autopct='%1.1f%%')
plt.title("Average score per hours studied")
plt.ylabel("Average score")
plt.show()
```

Does attendance affect exam scores?

On average, students attend 80-90% of their classes, and many of those students have scored higher than those who attended less; nevertheless, there are those who have attended over 90% of their classes and scored lower.

```python
plt.scatter(df["exam_score"], df["attendance_percent"], s=df["exam_score"]*30, alpha=0.6)
plt.title("Attendance vs. Exam scores")
plt.xlabel("Exam scores")
plt.ylabel("Attendance")
plt.show
```

Have exam scores stayed consistent throughout the years?

The current year's exam scores have dropped significantly compared to the previous year's. The average and minimum scores are much lower.

```python
import numpy as np

score_min=df["exam_score"].min()
score_q1=df["exam_score"].quantile(0.25)
print("Exam scores:", score_min, "Q1:", score_q1)

plt.figure(figsize=(7, 6))
df[["exam_score", "previous_scores"]].boxplot()
plt.title("Current and previous exam scores")
plt.ylabel("Value")
plt.show()
```
