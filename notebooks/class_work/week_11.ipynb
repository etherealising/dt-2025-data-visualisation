```python
sentence = "racecarisracecarbackwards"

# extract the first 3 letters
print("Print the first three letters:", sentence[0:3])
print(sentence[0:3])

# extract letters from 5-9
print(sentence[5:9])

# extract letters from 5 to the end; if you do not know how many letters there are until the end, then you can just leave it empty
print(sentence[5:])

# extract the last character; if you want to start counting backwards, then you have to add a minus in front
print(sentence[-1])

# extract the last four characters
print(sentence[-4:])

# extract the first letter of the substring starting from 5
print(sentence[5:][0])

# extract the first 5 letters of the substring starting from 5
print(sentence[5:][:5])

# extract entire string/text
print(sentence[:])

# extract every second character; same logic goes for every third character, every fourth...
print(sentence[::2])

# extract words from the sentence
print(sentence[0:7])
print(sentence[7:9])
print(sentence[9:16])
print(sentence[16:])
```

```python
sentence = "If I am a passenger princess, and you are also a passenger princess, then who is driving the car?"

words = sentence.split(" ") # split the sentence into separate words
print(words)

# extract the first word of the sentence
print(words[0])

# extract the first three words; adding an asterisk removes the square brackets
print(*words[0:3])

# extract the last three words
print(words[-3:])

# print number of words
print("The number of words: ", len(words))
```
